# ベイズ統計モデリングを用いた手法

- 共役事前分布：比較的単純なモデルでは事後分布を解析的に計算できる
    - 例えば分散既知の正規分布では、事後分布や予測分布も正規分布で簡単に計算できる
    - モデルが複雑になると、正規分布とガンマ分布を組み合わせた複雑な同時分布を考慮する必要がある

- MCMC（マルコフ連鎖モンテカルロ法）:解析的な計算が困難なモデルでもベイズ推定を実現する方法

### 解析的なベイズ推定による統計モデリングの実施フロー
- 共役事前分布や無情報事前分布を利用して、適切な事前分布を設定する
- ベイズの定理に基づき、パラメータの事後分布を解析的に計算する
- 推論データと事後分布から予測分布を求める

### MCMCによる統計モデリングの実施フロー
- ドメイン知識や無情報事前分布の知識をもとに、各パラメータに適切な事前分布を設定する
- MCMCにより、パラメータの事後分布に従うMCMCサンプルを得る。トレースプロットなどを用いて収束しているかを確認する
- 事後分布のMCMCサンプルを用いて、予測分布のMCMCサンプルを計算する
- 予測分布のMCMCサンプルにKDE（カーネル密度関数）を適用し、予測分布の確率密度関数を推定する
